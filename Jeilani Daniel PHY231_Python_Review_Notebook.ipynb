## Section 4: Strings

## Section 7: Range & For Loops

The `range()` function generates sequences of numbers. For loops let you repeat actions.

```python
print(list(range(5)))          # 0 to 4
print(list(range(5, 10)))      # 5 to 9
print(list(range(0, 10, 2)))   # 0 to 9, step by 2
print(list(range(10, 0, -1)))  # 10 down to 1
```

```python
# Loop through numbers
for i in range(5):
    print(i)
```

```python
# Loop through a list
for color in ["red", "green", "blue"]:
    print(color)
```

## Section 9: Functions

Functions let you reuse code. You define them with `def`.

```python
def say_hi():
    print("Just saying hello!")

say_hi()
```

```python
def say_it(message):
    print("Just saying: " + message)

say_it("Python is fun!")
```

```python
def compute_average_with_drop(grades):
    """Calculate average after dropping the lowest grade."""
    total = (sum(grades) - min(grades)) / (len(grades) - 1)
    return total

result = compute_average_with_drop([90, 78, 87, 10, 50])
print(result)
```

## BONUS Section 11: Visualization with Matplotlib

Matplotlib is Python's main plotting library. Here's a quick intro.

```python
import matplotlib.pyplot as plt

# Simple line plot
x = [1, 2, 3, 4, 5]
y = [2, 4, 6, 8, 10]

plt.plot(x, y)
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('My First Plot')
plt.show()
```

```python
# Bar chart
categories = ['A', 'B', 'C', 'D']
values = [23, 45, 12, 67]

plt.bar(categories, values)
plt.xlabel('Category')
plt.ylabel('Value')
plt.title('Simple Bar Chart')
plt.show()
```
